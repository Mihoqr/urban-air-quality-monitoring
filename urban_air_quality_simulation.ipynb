```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Number of simulated IoT readings
num_records = 100

# Data list for urban air quality monitoring
data = []

for _ in range(num_records):
    record = {
        "timestamp": datetime.now() - timedelta(minutes=np.random.randint(0, 1440)),  # Last 24 hours
        "station_id": f"Station_{np.random.choice(['A', 'B', 'C'])}",
        "PM2.5": round(np.random.uniform(5, 200), 1),  # PM2.5 µg/m³
        "CO2": round(np.random.uniform(300, 600), 1),  # CO2 ppm
        "temperature": round(np.random.uniform(15.0, 35.0), 1),  # Celsius
        "humidity": round(np.random.uniform(20.0, 80.0), 1),  # %
        "AQI": np.random.randint(0, 300)  # Simulated AQI index
    }
    data.append(record)

# Convert to DataFrame
df = pd.DataFrame(data)

# Save to files
df.to_csv("urban_air_quality_data.csv", index=False)
df.to_json("urban_air_quality_data.json", orient="records")

# Display first rows
df.head()
```

```python
import pandas as pd
import time

# Load CSV data (update the file name as needed)
df = pd.read_csv("urban_air_quality_data.csv")

def send_iot_data(device_id, pm25, co2, temperature, humidity, aqi):
    """Sends IoT sensor data to the deployed smart contract"""
    try:
        txn = contract.functions.storeData(
            device_id, pm25, co2, temperature, humidity, aqi
        ).transact({
            'from': web3.eth.default_account,
            'gas': 3000000
        })

        receipt = web3.eth.wait_for_transaction_receipt(txn)
        print(f"Data stored for {device_id}, Transaction Hash: {receipt.transactionHash.hex()}")
    except Exception as e:
        print(f"Failed to store data for {device_id}: {e}")

# Start sending data
print("Starting to send IoT data rows...\n")

print("Row count:", len(df))
print(df.head(), "\n")

for _, row in df.iterrows():
    print("Sending row:", row.to_dict())

    send_iot_data(
        str(row["station_id"]),               # device_id
        int(round(row["PM2.5"])),             # pm25
        int(round(row["CO2"])),               # co2
        int(round(row["temperature"])),       # temperature
        int(round(row["humidity"])),          # humidity
        int(round(row["AQI"]))                # aqi
    )

    time.sleep(1)

# Check results after sending all rows
total_records = contract.functions.getTotalRecords().call()
print(f"\nTotal IoT records stored: {total_records}")

record = contract.functions.getRecord(0).call()
print("First stored record:", record)

try:
    contract
except NameError:
    raise Exception("❌ Contract not defined. Please run the cell that sets up the smart contract.")

```
